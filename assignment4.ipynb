```python
#A function in Python is a self-contained block of code designed to perform a specific task.
#It is defined using the def keyword or using a lambda expression (for anonymous functions).
#Functions can accept inputs (called arguments), process them, and return a result using the return statement.
#They are not tied to any object or class and can be called independently from anywhere in the code.
# Function example
def greet(name):
    return "Hello " + name

print(greet("Abhinaba"))

#A method, on the other hand, is a function that is defined inside a class and is associated with objects created from that class.
#Methods are called using the dot (.) operator on the object and can access or modify the object’s internal state (its attributes).
#Since they belong to a class, they usually take self as the first parameter to refer to the instance of the object.
message = "hello"
print(message.upper())
#In this case, upper() is a method of the str class and works only on string objects.
```

```python
#Parameters are the variables listed inside the parentheses in the function definition.
#Arguments are the values passed to the function when it is called.
def greet(name):
    print("Hello", name)

greet("Abhinaba")
```

```python
# Using def
def greet(name="Abhinaba"):
    return f"Hello, {name}!"

print(greet())
print(greet("Pal"))

#Using args,kwargs
def add_all(*args):
    return sum(args)

print(add_all(1, 2, 3))

def show_info(**kwargs):
    return kwargs

print(show_info(name="Abhinaba", age=26))

#Using Lambda
square = lambda x: x * x
print(square(5))

#Using Nested function
def outer():
    def inner():
        return "Hello from inner!"
    return inner()

print(outer())

#Using function as variable
def say_hello():
    return "Hi!"

greeting = say_hello
print(greeting())

#Using function inside another function
def add(a, b):
    return a + b

def calculate():
    return add(3, 4)

print(calculate())

#Using map()
def double(x):
    return x * 2

print(list(map(double, [1, 2, 3])))
```

```python
#An iterable is any object that can be looped over, such as lists, tuples, strings, sets, and dictionaries.
#These objects implement the __iter__() method, which returns an iterator.
numbers = [1, 2, 3]
for num in numbers:  # `numbers` is an iterable
    print(num)


#An iterator, on the other hand, is an object that keeps track of the current position while iterating and implements both the __iter__() and __next__() methods.
#The key distinction is that while an iterable can give you a fresh iterator each time you loop through it, an iterator is a one-time-use object that gets exhausted once all its items are retrieved.
numbers = [1, 2, 3]
it = iter(numbers)  # now 'it' is an iterator

print(next(it))  # next(it) would now raise StopIteration
print(next(it))


#You can use a for loop on an iterable directly, but if you want to manually step through the elements one by one, you need an iterator and use the next() function.
#All iterators are inherently iterable because they have the __iter__() method, but not all iterables are iterators since they may lack __next__()
```

```python
#Generators are special functions that yield values one by one using the yield keyword instead of return.
def countdown(n):
    while n > 0:
        yield n
        n -= 1

# Taking user input
num = int(input("Enter a number to start countdown: "))

# Using the generator
for number in countdown(num):
    print(number)


#Advantags:
##Save memory (do not store all values at once)
##Faster for large data sets
```

```python
#A lambda function is a small anonymous function defined using the lambda keyword.
#It can have any number of arguments but only one expression.
#The result of that expression is automatically returned.

#Lambda is typically used when the logic is simple, used once, and readability is not sacrificed. Otherwise, go with def.

#filter
nums = [1, 2, 3, 4]
even = list(filter(lambda x: x % 2 == 0, nums))
print(even)
#sort
names = ["Rahul", "Ankit", "Sourav"]
sorted_names = sorted(names, key=lambda x: (len(x), x))
print(sorted_names)
```

```python
#map() applies a function to every item in an iterable.
nums = [1, 2, 3]
squares = list(map(lambda x: x*x, nums))
print(squares)

#differences:
##map(): Transforms each item in an iterable using a function.
##reduce(): Combines all items in an iterable into a single value using a function.
##filter(): Selects items from an iterable that meet a condition.

nums = [1,2,3,4]

# map
print(list(map(lambda x: x*2, nums)))
# reduce
print(reduce(lambda x, y: x + y, nums))
# filter
print(list(filter(lambda x: x%2==0, nums)))
```

```python
class Fibonacci:
    def __init__(self, n):
        self.n = n
        self.a, self.b = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n:
            raise StopIteration
        self.count += 1
        result = self.a
        self.a, self.b = self.b, self.a + self.b
        return result
n = int(input("Enter how many Fibonacci numbers you want: "))
for num in Fibonacci(n):
    print(num, end=" ")
```

```python
def powers_of_two(max_exp):
    for i in range(max_exp + 1):
        yield 2 ** i
n = int(input("Enter the maximum exponent: "))
print(f"Powers of 2 from 2^0 to 2^{n}:")
for power in powers_of_two(n):
    print(power, end=" ")
```

```python
def read_lines(file_name):
    try:
        with open(file_name, 'r') as file:
            for line in file:
                yield line.strip()  # Remove extra whitespace or newlines
    except FileNotFoundError:
        print("❌ File not found. Please check the file name and path.")

# Ask the user for the file name
file_name = input("Enter the file name to read: ")

# Use the generator to print each line
if file_name:
    for line in read_lines(file_name):
        print(line)
```

```python
# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32
celsius_temps_input = input("Enter temperatures in Celsius separated by spaces: ")
celsius_temps = list(map(int, celsius_temps_input.split()))
fahrenheit_temps = list(map(celsius_to_fahrenheit, celsius_temps))
print("Celsius temperatures:", celsius_temps)
print("Fahrenheit temperatures:", fahrenheit_temps)
```

```python
# Function to filter vowels from given strng
def is_not_vowel(char):
    return char.lower() not in 'aeiou'
input_string = input("Enter a string: ")
filtered_string = ''.join(filter(is_not_vowel, input_string))
print("String without vowels:", filtered_string)
```
